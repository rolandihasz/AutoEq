# Import and init

```python
from glances.main import GlancesMain
from glances.stats import GlancesStats
```

```python
core = GlancesMain()
```

```python
stats = GlancesStats(config=core.get_config(), args=core.get_args())
```

# Get availables plugins list

```python
stats.getPluginsList()
```

# Update the stats

```python
stats.update()
```

# Get the stats (example for the CPU plugin)

The ***get_raw()*** method returns a Python object (a dict for CPU plugin):

```python
stats.get_plugin('cpu').get_raw()
```

For a JSON compliant object use ***get_stats()***:

```python
stats.get_plugin('cpu').get_stats()
```

# Get stats history (system and user are historised for the CPU plugin)

First of all, we generate some stats:

```python
import time
for i in range(10):
  stats.get_plugin('cpu').update()
  time.sleep(1)
```

Get stats history for CPU. It will return a dict of list. Each item of the list is a set with the date and the value:

```python
stats.get_plugin('cpu').get_raw_history()
```

It is also possible to get history in a JSON compliant format:

```python
stats.get_plugin('cpu').get_stats_history()
```
